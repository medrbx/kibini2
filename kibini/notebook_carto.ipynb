```python
import pandas as pd
test = pd.read_excel("../data/aecs/tableaux_suivi_AECS_20250917.xlsx",
                     sheet_name='menus_déroulants',
                     usecols="T:AE",
                     skiprows=1 #Passer une ligne dans la lecture du fichier
                    )
test
```

```python
établissements = []
dico = {}
for i in test:
    établissements.append(i)
    for v in test[i].dropna():
        établissements.append(i + ' ' + v)
dico["Liste d'établissements"] = établissements
établissements
```

```python
df_établissements = pd.DataFrame(dico)
df_établissements
```
