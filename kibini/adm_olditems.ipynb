```python
import pandas as pd
from datetime import datetime as dt
from datetime import date

from kiblib.utils.db import DbConn
```

```python
db_conn = DbConn().create_engine()
```

```python
query = """SELECT itemnumber, biblionumber, barcode, dateaccessioned, booksellerid, homebranch, price, replacementprice, replacementpricedate, datelastborrowed, datelastseen, stack, notforloan, damaged, damaged_on, itemlost, itemlost_on, withdrawn, withdrawn_on, itemcallnumber, holdingbranch, `timestamp`, location, onloan, ccode
FROM koha_prod.deleteditems;"""

deleteditems = pd.read_sql(query,db_conn)

query = """SELECT itemnumber, biblionumber, barcode, dateaccessioned, booksellerid, homebranch, price, replacementprice, replacementpricedate, datelastborrowed, datelastseen, stack, notforloan, damaged, damaged_on, itemlost, itemlost_on, withdrawn, withdrawn_on, itemcallnumber, holdingbranch, `timestamp`, location, onloan, ccode
FROM koha_prod.items;"""

items = pd.read_sql(query,db_conn)

allitems = pd.concat([items, deleteditems])
len(allitems)
```

```python
query = """SELECT biblionumber, itemtype FROM koha_prod.deletedbiblioitems;"""

deletedbiblioitems = pd.read_sql(query,db_conn)

query = """SELECT biblionumber, itemtype FROM koha_prod.biblioitems;"""

biblioitems = pd.read_sql(query,db_conn)

allbiblioitems = pd.concat([biblioitems, deletedbiblioitems])
len(biblioitems)
```

```python
query = """SELECT biblionumber, title FROM koha_prod.deletedbiblio;"""

deletedbiblio = pd.read_sql(query,db_conn)

query = """SELECT biblionumber, title FROM koha_prod.biblio;"""

biblio = pd.read_sql(query,db_conn)

allbiblio = pd.concat([biblio, deletedbiblio])
len(biblio)
```
