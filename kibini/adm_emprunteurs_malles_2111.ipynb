```python
import pandas as pd
from datetime import datetime as dt
from datetime import date

from kiblib.utils.db import DbConn
```

```python
db_conn = DbConn().create_engine()
```

```python
query = """
SELECT
    ExtractValue(i.more_subfields_xml,'//datafield[@tag="999"]/subfield[@code="8"]') as code_malle,
    i.itemnumber,
    i.barcode,
    b.cardnumber 
FROM koha_prod.items i
JOIN koha_prod.issues iss ON i.itemnumber = iss.itemnumber
JOIN koha_prod.borrowers b ON b.borrowernumber = iss.borrowernumber 
WHERE ExtractValue(i.more_subfields_xml,'//datafield[@tag="999"]/subfield[@code="8"]') != ''
ORDER BY code_malle
"""

ex_malles = pd.read_sql(query,db_conn)
ex_malles
```
