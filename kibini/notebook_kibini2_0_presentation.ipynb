#  LES TABLEAUX DE BORDS DE LA GRAND PLAGE

## C'est quoi ? 📊

Les tableaux de bord de La Grand-Plage est un outil pensé pour **suivre l'activité quotidienne de la Médiathèque de Roubaix.** Créé en 2016, il a fait l'objet d'une refonte technique et graphique.  



## Ca sert à quoi ? 🧭

Comme son nom l'indique, **<span style="color:blue">les tableaux de bords La Grand-Plage</span>** permet à la Direction de la Médiathèque (mais aussi les collègues intéressés !!!) **de mieux comprendre l'activité de la médiathèque dans les grandes lignes**. Elle a été pensé comme un outil qui permet de répondre à des questions simples en se basant sur une arborescence qui reprend les grands indicateurs statistiques de la Médiathèque de Roubaix.

Le site se compose d'un menu (situé à gauche) composé de questions auxquels les graphiques présents dans chaque page doivent apporter des éléments de réponses.

## D'où proviennent les données des graphiques ? 🔎

Les données proviennent de 3 grandes sources : 
* **Données du SIGB Koha :** données concernant les usagers et les collections
* **Données des collègues :** données concernant les actions culturelles, éducatives et sociales
* **Données externes :** données concernant les ordinateurs, le wifi, le portail documentaire, la bn-r et les entrées


## Comment ça fonctionne ? ⚙️

Les tableaux de bords utilise plusieurs briques pour fonctionner : 
* **PERL :** pour les scripts qui aliment les bases de données (sources de données)
* **SQL :** pour l'hébergement des données et les requêtes basiques
* **Python :** pour le traitement des données et la génération de graphiques
* **Dancer :** pour la construction du site Kibini qui contient les tableaux de bords de La Grand-Plage 
* **HTML : :** pour l'affichage des pages graphiques**


## Comment l'utiliser ? 🤔

Le menu peut-être divisé en 3 grandes zones :
1. **Les données générales** tous sites confondus
2. **Les données par sites** (Médiathèque, Zèbre)
3. **Les grands tableaux de la Politique Documentaire (Poldoc)** utilisés lors de l'analyse annuelle des collections

## On a besoin de vous ! 💡

Les données propres ça n'existe pas ! Mais vous pouvez quand même aider à améliorer notre outil en :
* **Remontant les erreurs ou données** "anormales" ; ça nous permet de corriger/améliorer les scripts
* **Proposant des idées** qui répondent à vos besoin d'informations (contacer Lucas & François☺)

## Récap des services de la Médiathèque

La Médiathèque est ouverte 54h par semaine en période scolaire, de 9h à 19h du mardi au samedi, et le dimanche de 14h à 18h hors période de vacances scolaires. Elle donne accès à :
* un espace de 3350 m²
* 275 places assises (+ 50 places avec ordinateurs),
* Une salle d'étude silencieuse,
* Une salle informatique (Salle Multimédia),
* 8 imprimantes noir et blanc et couleur, 2 scanners et 1 photocopieuse,
* Au total, 50 postes informatiques accessibles au public,
* 40 prises pour connecter son ordinateur personnel,
* 185 000 documents empruntables et près de 60 000 en consultation sur place,
* Livres, revues et journaux, bandes dessinées, livres sonores, livres en gros caractères, CD, DVD, méthodes de langue, partitions
* Chaque semaine, des ateliers CV, des permanences d'écrivain public, de l'aide aux devoirs,
* Une salle équipée et modulable, La Criée, pour accueillir lectures, expositions, concerts, projections, ...
* Des rendez-vous réguliers pour des publics variés (Les Racontées, Les ateliers Clics et Déclics, le Club de lecture Tulalus, les 15h tympantes, les Ateliers du Patrimoine, les Questions de Parents, ...)
* Et une large programmation culturelle pour petits et grands, toujours renouvelée
