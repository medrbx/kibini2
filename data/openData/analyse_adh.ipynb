```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
adh = pd.read_csv("data/adherents.csv")
adh['annee'] = adh['date_extraction'].str[:4]

adh = adh[adh['annee'] != '2020']
```

## Evolution inscrits

```python
group = adh.groupby(['annee'])['date_extraction'].count().reset_index()
group = group.rename(columns={'date_extraction':'inscrits'})
group
```

```python
sns.barplot(data=group, x='annee', y='inscrits')
```

## Fréquences de venue

```python
# part des personnes n'ayant laissé aucune trace
var = 'nb_venues'
data = pers[pers[var] == 0]

group1 = pers.groupby(['annee'])['date_extraction'].count().reset_index()
group1 = group1.rename(columns={'date_extraction':'inscrits'})

group2 = data.groupby(['annee'])['date_extraction'].count().reset_index()
group2 = group2.rename(columns={'date_extraction':'pas venus'})

group = group1.merge(group2, on='annee')
group['part pas venus'] = round(group['pas venus'] / group['inscrits'] * 100, 0)
group
```

```python
# part des personnes inscrites dans l'année n'ayant laissé aucune trace
var = 'nb_venues'
data = pers[(pers[var] == 0) & (pers['inscription_fidelite'] == 0)]

group1 = pers[(pers['inscription_fidelite'] == 0)].groupby(['annee'])['date_extraction'].count().reset_index()
group1 = group1.rename(columns={'date_extraction':'inscrits'})

group2 = data.groupby(['annee'])['date_extraction'].count().reset_index()
group2 = group2.rename(columns={'date_extraction':'pas venus'})

group = group1.merge(group2, on='annee')
group['part pas venus'] = round(group['pas venus'] / group['inscrits'] * 100, 0)
group
```
