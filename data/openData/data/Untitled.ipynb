```python
adh23 = pd.read_csv(f"adherents_2023_brut.csv")
adh23['activite'] = ''

adh23['nb_venues_prets'] = adh23['nb_venues_prets_bus'] + adh23['nb_venues_prets_mediatheque']

adh23['activite_emprunteur'] = 'Non emprunteur'
adh23.loc[ adh23['nb_venues_prets'] > 0, 'activite_emprunteur'] = 'Emprunteur'
adh23.loc[ adh23['nb_venues_prets'] > 0, 'activite'] = adh23['activite'] + 'prêt'


adh23['activite_emprunteur_bus'] = 'Non emprunteur Zèbre'
adh23.loc[ adh23['nb_venues_prets_bus'] > 0, 'activite_emprunteur_bus'] = 'Emprunteur Zèbre'

adh23['activite_emprunteur_med'] = 'Non emprunteur Médiathèque'
adh23.loc[ adh23['nb_venues_prets_mediatheque'] > 0, 'activite_emprunteur_med'] = 'Emprunteur Médiathèque'

adh23['activite_salle_etude'] = "Non utilisateur Salle d'étude"
adh23.loc[ adh23['nb_venues_salle_etude'] > 0, 'activite_salle_etude'] = "Utilisateur Salle d'étude"
adh23.loc[ adh23['nb_venues_salle_etude'] > 0, 'activite'] = adh23['activite'] + ' étude'

adh23['activite_utilisateur_postes_informatiques'] = 'Non utilisateur postes informatiques'
adh23.loc[ adh23['nb_venues_postes_informatiques'] > 0, 'activite_utilisateur_postes_informatiques'] = "Utilisateur postes informatiques"
adh23.loc[ adh23['nb_venues_postes_informatiques'] > 0, 'activite'] = adh23['activite'] + ' postes'


adh23['activite_utilisateur_wifi'] = 'Non utilisateur Wifi'
adh23.loc[ adh23['nb_venues_wifi'] > 0, 'activite_utilisateur_wifi'] = "Utilisateur Wifi"
adh23.loc[ adh23['nb_venues_wifi'] > 0, 'activite'] = adh23['activite'] + ' wifi'


#adh23['activite'] = 'aucune trace'
adh23['activite'] = adh23['activite'].str.strip()
adh23['activite'] = adh23['activite'].str.replace(" ", " + ")

adh23.loc[ adh23['activite'] == '', 'activite'] = 'aucune activité'
```

```python
adh23 =adh23.rename(columns={"adh_age_lib1":"ages_lib1",
                             "adh_age_lib2":"ages_lib2",
                             "adh_age_lib3":"ages_lib3",
                             "adh_geo_gentile": "geo_gentile",
                             })
```
