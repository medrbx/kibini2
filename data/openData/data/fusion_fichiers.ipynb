```python
import pandas as pd
import json
```

```python
adh_all = pd.DataFrame()
for year in [2022, 2021, 2020, 2019, 2018, 2017, 2016, 2015]:
    adh = pd.read_csv(f"adherents_{year}.csv")
    adh_all = pd.concat([adh_all, adh])
```

```python
adh_all.columns
```

```python
adh_all['inscription_attribut'].value_counts()
```

```python
def isjson(x):
    x = str(x)
    if x[0] == '{':
        return json.loads(x)
adh_all['inscription_attribut_json'] = adh_all['inscription_attribut'].apply(isjson)
```

```python
keys = []
for x in adh_all['inscription_attribut_json']:
    if x:
        keys.extend(x.keys())
keys = list(dict.fromkeys(keys))
keys
```

```python
def json2action(x):
    if x:
        if 'action' in x:
            if isinstance(x['action'], list):
                return x['action'][0]
            else:
                return x['action']
adh_all['inscription_attribut_action'] = adh_all['inscription_attribut_json'].apply(json2action)

def json2zebre(x):
    if x:
        if 'zÃ¨bre' in x:
            if isinstance(x['zÃ¨bre'], list):
                return x['zÃ¨bre'][0]
            else:
                return x['zÃ¨bre']
adh_all['inscription_attribut_zèbre'] = adh_all['inscription_attribut_json'].apply(json2zebre)

def json2coll(x):
    if x:
        if 'collectivitÃ©s' in x:
            if isinstance(x['collectivitÃ©s'], list):
                return x['collectivitÃ©s'][0]
            else:
                return x['collectivitÃ©s']
adh_all['inscription_attribut_collectivités'] = adh_all['inscription_attribut_json'].apply(json2coll)

def json2pcs(x):
    if x:
        if 'PCS' in x:
            if isinstance(x['PCS'], list):
                return x['PCS'][0]
            else:
                return x['PCS']
adh_all['inscription_attribut_pcs'] = adh_all['inscription_attribut_json'].apply(json2pcs)
```

```python
adh_all.to_csv("adherents.csv", index=False)
```
